# Experiment 1
## Generating Random Samples from Probability Distributions
**Aim:** Generate random samples from Normal and Uniform distributions and visualize them.

### Step 1 – Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Step 2 – Generate Samples from Normal Distribution

```python
mu, sigma = 0, 1          # mean and standard deviation
normal_samples = np.random.normal(mu, sigma, 1000)
print('Normal Samples (first 10):', normal_samples[:10].round(3))
```

### Step 3 – Generate Samples from Uniform Distribution

```python
low, high = 0, 1
uniform_samples = np.random.uniform(low, high, 1000)
print('Uniform Samples (first 10):', uniform_samples[:10].round(3))
```

### Step 4 – Generate Samples from Exponential Distribution

```python
lam = 1.5
exponential_samples = np.random.exponential(1/lam, 1000)
print('Exponential Samples (first 10):', exponential_samples[:10].round(3))
```

### Step 5 – Visualize All Three Distributions

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(normal_samples, bins=30, color='steelblue', edgecolor='black')
axes[0].set_title('Normal Distribution (μ=0, σ=1)')
axes[0].set_xlabel('Value'); axes[0].set_ylabel('Frequency')

axes[1].hist(uniform_samples, bins=30, color='orange', edgecolor='black')
axes[1].set_title('Uniform Distribution [0, 1]')
axes[1].set_xlabel('Value'); axes[1].set_ylabel('Frequency')

axes[2].hist(exponential_samples, bins=30, color='green', edgecolor='black')
axes[2].set_title('Exponential Distribution (λ=1.5)')
axes[2].set_xlabel('Value'); axes[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Step 6 – Print Basic Statistics

```python
for name, samples in [('Normal', normal_samples), ('Uniform', uniform_samples), ('Exponential', exponential_samples)]:
    print(f'{name}: Mean={samples.mean():.3f}, Std={samples.std():.3f}, Min={samples.min():.3f}, Max={samples.max():.3f}')
```

### Result
Random samples were successfully generated from Normal, Uniform, and Exponential distributions and visualized using histograms.
