## Evaluation Metrics for Generative Models
**Aim:** Implement and demonstrate common evaluation metrics for classification and generative models.

### Step 1 – Import Libraries

```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
```

### Step 2 – Simulate True Labels and Predicted Labels

```python
np.random.seed(42)
y_true = np.random.randint(0, 2, 100)          # 0 or 1 (binary)
y_pred = np.random.randint(0, 2, 100)
print('True labels (first 10):', y_true[:10])
print('Pred labels (first 10):', y_pred[:10])
```

### Step 3 – Compute Classification Metrics

```python
acc  = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred)
rec  = recall_score(y_true, y_pred)
f1   = f1_score(y_true, y_pred)

print(f'Accuracy : {acc:.4f}')
print(f'Precision: {prec:.4f}')
print(f'Recall   : {rec:.4f}')
print(f'F1-Score : {f1:.4f}')
```

### Step 4 – Confusion Matrix

```python
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(5, 4))
plt.imshow(cm, cmap='Blues')
plt.title('Confusion Matrix')
plt.colorbar()
plt.xlabel('Predicted'); plt.ylabel('Actual')
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i,j], ha='center', va='center', color='black', fontsize=14)
plt.xticks([0,1], ['Negative','Positive']); plt.yticks([0,1], ['Negative','Positive'])
plt.tight_layout(); plt.show()
```

### Step 5 – Compute Perplexity (for generative models)
Perplexity measures how well a probability model predicts a sample.

```python
# Simulated log-probabilities from a language/generative model
log_probs = np.random.uniform(-3, -0.5, 50)   # log p(x) values
perplexity = np.exp(-np.mean(log_probs))
print(f'Perplexity: {perplexity:.4f}')
```

### Step 6 – Compute Mean Squared Error and SSIM (image quality metrics)

```python
from skimage.metrics import structural_similarity as ssim
img1 = np.random.rand(64, 64)
img2 = img1 + np.random.normal(0, 0.1, (64, 64))   # add small noise
mse  = np.mean((img1 - img2)**2)
ssim_val = ssim(img1, img2, data_range=img2.max()-img2.min())
print(f'MSE : {mse:.4f}')
print(f'SSIM: {ssim_val:.4f}  (1.0 = identical)')
```

### Result
Key evaluation metrics — accuracy, precision, recall, F1, confusion matrix, perplexity, MSE, and SSIM — were successfully computed and demonstrated.
