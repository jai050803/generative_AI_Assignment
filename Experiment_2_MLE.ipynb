# Experiment 2
## Maximum Likelihood Estimation (MLE)
**Aim:** Implement MLE to estimate parameters of a Normal distribution from a synthetic dataset.

### Step 1 – Import Libraries

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
```

### Step 2 – Generate Synthetic Dataset with Known Parameters

```python
true_mu, true_sigma = 5.0, 2.0
np.random.seed(42)
data = np.random.normal(true_mu, true_sigma, 500)
print(f'True Parameters  → μ = {true_mu}, σ = {true_sigma}')
```

### Step 3 – Compute MLE Estimates
For a Normal distribution, MLE gives: **μ_mle = mean(data)** and **σ_mle = std(data)**

```python
mle_mu    = np.mean(data)
mle_sigma = np.std(data)          # MLE uses N in denominator
print(f'MLE Estimates    → μ = {mle_mu:.4f}, σ = {mle_sigma:.4f}')
```

### Step 4 – Calculate Estimation Error

```python
error_mu    = abs(true_mu    - mle_mu)
error_sigma = abs(true_sigma - mle_sigma)
print(f'Estimation Error → Δμ = {error_mu:.4f}, Δσ = {error_sigma:.4f}')
```

### Step 5 – Visualize Data vs Fitted Distribution

```python
x = np.linspace(data.min(), data.max(), 200)
plt.figure(figsize=(8, 4))
plt.hist(data, bins=30, density=True, alpha=0.6, color='steelblue', label='Observed Data')
plt.plot(x, norm.pdf(x, mle_mu, mle_sigma), 'r-', lw=2, label=f'MLE Fit (μ={mle_mu:.2f}, σ={mle_sigma:.2f})')
plt.plot(x, norm.pdf(x, true_mu, true_sigma), 'g--', lw=2, label=f'True Dist (μ={true_mu}, σ={true_sigma})')
plt.legend(); plt.title('MLE Fitting'); plt.xlabel('Value'); plt.show()
```

### Result
MLE successfully estimated the parameters of the Normal distribution. The estimated values are very close to the true parameters.
